````python
def get_completion(client, input, model = "gpt-5.6-luna"):
    response = client.responses.create(
        model=model,
        input=input,
        store=True,
    )

    return response.output_text
````

````python
import os
from openai import OpenAI


client = OpenAI(
  api_key=os.getenv("OPENAI_API_KEY")
)

print(get_completion(client, "How are you"))
````
